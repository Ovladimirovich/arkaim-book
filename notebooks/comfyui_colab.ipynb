## 4. Start ComfyUI

```python
import subprocess, time

comfyui_proc = subprocess.Popen(
    ['python', 'main.py', '--listen', '0.0.0.0', '--port', '8188'],
    cwd='/content/ComfyUI',
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT
)

print('Starting ComfyUI...')
time.sleep(20)

import requests
try:
    r = requests.get('http://localhost:8188/system_stats')
    print(f'ComfyUI running! {r.json()}')
except Exception as e:
    print(f'Still starting ({e}), waiting 10s...')
    time.sleep(10)
    r = requests.get('http://localhost:8188/system_stats')
    print(f'ComfyUI running! {r.json()}')
```

## 5. Cloudflare Tunnel

Quick Tunnel (no account, URL changes on restart).

```python
import subprocess, re

tunnel_proc = subprocess.Popen(
    ['cloudflared', 'tunnel', '--url', 'http://localhost:8188'],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

url = None
print('Waiting for tunnel URL...')
for line in iter(tunnel_proc.stdout.readline, ''):
    print(line.strip())
    match = re.search(r'https://[a-zA-Z0-9-]+\.trycloudflare\.com', line)
    if match:
        url = match.group(0)
        break

if url:
    print(f'\n{"="*60}')
    print(f' ComfyUI Tunnel URL: {url}')
    print(f'{"="*60}')
    print(f'\nCopy to .env: COMFYUI_URL={url}')
    with open('/content/comfyui_url.txt', 'w') as f:
        f.write(url)
    print('URL saved to /content/comfyui_url.txt')
else:
    print('Failed to get URL')
```

## 6. Verify

```python
import requests

try:
    with open('/content/comfyui_url.txt') as f:
        url = f.read().strip()
except FileNotFoundError:
    print('No URL found. Run cell 5.')
    url = None

if url:
    r = requests.get(f'{url}/system_stats', timeout=30)
    print(f'Status: {r.status_code}')
    print(f'Stats: {r.json()}')
    print(f'\nAll working! Set COMFYUI_URL={url} in .env')
```
